```python
import pandas as pd
import pyBigKinds as pbk
from bertopic import BERTopic
from sentence_transformers import SentenceTransformer
```

```python
sentence_model = SentenceTransformer('jhgan/ko-sroberta-multitask')
topic_model = BERTopic.load("./bin/bipolar_model.bin", embedding_model=sentence_model)
```

```python
topic_model.visualize_hierarchy()
```

```python
def list_to_str(words: list):
    """Function that list data change to string for text preprocessing"""
    for i in range(len(words)):
        text = ""
        for word in words[i]:
            if text == "":
                text = word
            else:
                text = text + " " + word
        words[i] = text
    return words

df = pd.read_excel("./bipolar_disorder/df_bipolar.xlsx", engine="openpyxl")

words = pbk.keyword_parser(pbk.keyword_list(df))
words = list_to_str(words)

hierarchical_topics = topic_model.hierarchical_topics(words)
```

```python
topic_model.visualize_hierarchy(hierarchical_topics=hierarchical_topics)
```
