```python
import glob
import pandas as pd
import re
from tqdm import tqdm
from konlpy.tag import Mecab

import warnings
warnings.filterwarnings("ignore")
```

```python
lis = glob.glob('1960~1999 네이버 뉴스 라이브러리/txt/*.txt')
df = pd.DataFrame(columns=["제목","언론사","일자","키워드"])


for name in lis:
    name_list = name.split("_")
    for _ in range(len(name_list[:-3])):
        name_list[0] = name_list[0] + "_" + name_list[1]
        del(name_list[1]) 
    article = name_list[0].split("/")[2]
    press = name_list[1].split('1')[0]
    date = re.sub("[^0-9]" ,"", name_list[1])
    f = open(name, 'r')
    text = str(f.readlines()[2:])
    tmp = pd.DataFrame({"제목": [article], 
                       "언론사": [press], 
                       "일자": [date],
                       "키워드": [text], 
                       })
    df = pd.concat([df, tmp])
    
df.reset_index(inplace=True, drop=True)
df.head(10)
```

```python
lis = glob.glob('1965~1989 중앙일보/*.txt')
jungang_df = pd.DataFrame(columns=["제목", "언론사", "일자", "키워드"])

for name in lis:
    name_list = name.split("-")
    article = name_list[0].split("/")[1]
    article = re.sub("_", " ", article)
    press = "중앙일보"
    date = name_list[1].split("_")[0]
    date = re.sub("[^0-9]", "", date)
    f = open(name, 'r')
    text = f.read()
    tmp = pd.DataFrame({"제목": [article], 
                       "언론사": [press], 
                       "일자": [date],
                       "키워드": [text], 
                       })
    jungang_df = pd.concat([jungang_df, tmp])
    
jungang_df.reset_index(inplace=True, drop=True)
jungang_df.head(10)
```

```python
lis = glob.glob('2000~2024 조선일보/txt/*.txt')
xlsx = pd.read_excel("2000~2024 조선일보/우울증_조선일보_20180101-20240930.xlsx", engine="openpyxl")[["제목","언론사","일자","키워드"]]
josun_df = pd.DataFrame(columns=["제목","언론사","일자","키워드"])


for name in lis:
    name_list = name.split("-")
    if name_list[1][0].isdigit()==False:
        name_list[0] = name_list[0] + "_" + name_list[1]
        del(name_list[1])
    article = name_list[0].split("/")[2]
    press = "조선일보"
    date = name_list[1].split("_")[0]
    if date[0].isdigit()==False:
        date = name_list[-1].split("_")[0]
        article = article + name_list[1].split("_")[0]
    date = re.sub("[^0-9]", "", date)
    article = re.sub("_", " ", article)
    f = open(name, 'r')
    text = f.read()
    tmp = pd.DataFrame({"제목": [article], 
                    "언론사": [press], 
                    "일자": [date],
                    "키워드": [text], 
                    })
    josun_df = pd.concat([josun_df, tmp])
    
josun_df = pd.concat([josun_df, xlsx])
    
josun_df.reset_index(inplace=True, drop=True)
josun_df.head(10)
```

```python
bigkinds_df = pd.DataFrame()
lis = glob.glob('1990~2024 빅카인즈/*.xlsx')

for data in lis:
    tmp = pd.read_excel(data, engine="openpyxl")
    tmp = tmp[["제목","언론사","일자","키워드"]]
    bigkinds_df = pd.concat([bigkinds_df, tmp])
    
bigkinds_df.drop_duplicates(inplace=True)
bigkinds_df.reset_index(inplace=True, drop=True)
bigkinds_df.tail(20)
```

```python
df = pd.concat([df, jungang_df])
df = pd.concat([df, josun_df])
df = pd.concat([df, bigkinds_df])
df.info()
```

```python
df["text"] = df["키워드"].str.replace("[^가-힣|0-9]","", regex=True)
df.head(10)
```

```python
def tokenize(sentence:pd.Series):

    tokenizer = Mecab()
    tokenized=[]
    temp = tokenizer.nouns(sentence)
    temp = [word for word in temp] 
    tokenized.append(temp)
    
    for token in tokenized:
        string = ""
        for num in range(len(token)):
            if (len(token[num])>=2) & (num != len(token)-1):
                string += token[num] + ","
            elif num == len(token)-1:
                string += token[num]
    return string
```

```python
df["키워드"] = df["text"].apply(tokenize)
df.drop("text", axis=1, inplace=True)
```
