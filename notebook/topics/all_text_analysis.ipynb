```python
import pandas as pd
from pyBigKinds.preprocessing import tfidf
import matplotlib.pyplot as plt

import os
import warnings
warnings.filterwarnings("ignore")

os.chdir('../../model')

bipolar_all= pd.read_excel("bipolar_disorder/df_bipolar.xlsx", engine="openpyxl")

schizo_all = pd.read_excel("schizophrenia/df_schizo.xlsx", engine="openpyxl")

depression_all = pd.read_excel("depression/df_depression.xlsx", engine="openpyxl")

plt.rcParams["font.family"] = "NanumGothic"
plt.rcParams['axes.unicode_minus'] = False

pd.options.display.float_format = '{:.2f}'.format
```

```python
def t_tfidf(df):
    tmp = df.copy()
    result = tfidf(tmp)
    result["점수"] = result["빈도"].apply(lambda x: round(x,2))
    result = result.drop("빈도", axis=1)
    return result
```

```python
tmp1 = t_tfidf(bipolar_all)
tmp2 = t_tfidf(schizo_all)
tmp3 = t_tfidf(depression_all)

tmp1.columns = ["양극성 장애 단어", "양극성 장애 상대 빈도"]
tmp2.columns = ["조현병 단어", "조현병 상대 빈도"]
tmp3.columns = ["우울증 단어", "우울증 상대 빈도"]

data = pd.concat([tmp1, tmp2, tmp3], axis = 1).head(20).style.set_caption("정신질환 별 TF-IDF 상대 빈도 상위 20개")
data
```
